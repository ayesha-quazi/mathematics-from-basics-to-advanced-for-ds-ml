# Linear Algebra Basics for Machine Learning

This notebook demonstrates fundamental linear algebra operations that are essential for understanding machine learning algorithms.

## Topics Covered
1. Vectors and Vector Operations

```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)
```

## 1. Vector Operations

Vectors are fundamental data structures in ML. Each data point can be represented as a vector.

```python
# Create two vectors
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

print(f"v1 = {v1}")
print(f"v2 = {v2}")
print(f"\nv1 + v2 = {v1 + v2}")
print(f"v1 · v2 = {np.dot(v1, v2)}")
print(f"||v1|| = {np.linalg.norm(v1):.4f}")
```

### Visualizing Vectors

```python
# Visualize 2D vectors
fig, ax = plt.subplots(figsize=(8, 6))

origin = [0, 0]
v1_2d = np.array([3, 2])
v2_2d = np.array([1, 4])

ax.quiver(*origin, *v1_2d, color='blue', scale=1, scale_units='xy', angles='xy', label='v1')
ax.quiver(*origin, *v2_2d, color='red', scale=1, scale_units='xy', angles='xy', label='v2')
ax.quiver(*origin, *(v1_2d + v2_2d), color='green', scale=1, scale_units='xy', angles='xy', label='v1 + v2')

ax.set_xlim(-1, 5)
ax.set_ylim(-1, 7)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Vector Addition')
plt.show()
```

## 2. Matrix Operations

Matrices are used to represent datasets, transformations, and model parameters.

```python
# Create matrices
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8],
              [9, 10],
              [11, 12]])

print("Matrix A (2x3):")
print(A)
print(f"\nMatrix B (3x2):")
print(B)
print(f"\nA @ B (matrix multiplication, 2x2):")
print(A @ B)
```

## 3. Solving Linear Systems

Many ML algorithms require solving systems of linear equations.

```python
# System: 2x + y = 5
#         x + 3y = 11

A = np.array([[2, 1],
              [1, 3]])
b = np.array([5, 11])

x = np.linalg.solve(A, b)
print(f"Solution: x = {x[0]:.2f}, y = {x[1]:.2f}")
print(f"Verification: A @ x = {A @ x}")
print(f"Expected: b = {b}")
```

## 4. Application: Linear Regression

Linear regression uses the normal equation: θ = (XᵀX)⁻¹Xᵀy

```python
# Generate synthetic data
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X.squeeze() + np.random.randn(100)

# Add bias term
X_with_bias = np.c_[np.ones((100, 1)), X]

# Normal equation
theta = np.linalg.inv(X_with_bias.T @ X_with_bias) @ X_with_bias.T @ y

print(f"True parameters: y = 4 + 3x")
print(f"Learned parameters: y = {theta[0]:.2f} + {theta[1]:.2f}x")
```

```python
# Visualize the results
plt.figure(figsize=(10, 6))
plt.scatter(X, y, alpha=0.5, label='Data')
X_plot = np.array([[0], [2]])
X_plot_with_bias = np.c_[np.ones((2, 1)), X_plot]
y_plot = X_plot_with_bias @ theta
plt.plot(X_plot, y_plot, 'r-', linewidth=2, label=f'Fit: y={theta[0]:.2f}+{theta[1]:.2f}x')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression using Normal Equation')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
